```python
from sklearn.svm import SVR
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score
from sklearn import datasets
```

```python
df = datasets.load_diabetes(as_frame=True).frame
```

```python
X = df.drop("target", axis=1)
y = df["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Scaling

y_scaler = StandardScaler()

y_train_scaled = y_scaler.fit_transform(y_train.values.reshape(-1,1)).ravel()
y_test_scaled = y_scaler.fit_transform(y_test.values.reshape(-1,1)).ravel()
```

```python
# Model

model = SVR()
model.fit(X_train, y_train_scaled)
```

```python
y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)
```

```python
print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# Linear kernel

model = SVR(kernel="linear")
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# Polynomial kernel

model = SVR(kernel="poly")
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```

```python
# Sigmoid kernel

model = SVR(kernel="sigmoid")
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```

# HyperParameter Tuning Using GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "C": [1, 2, 5, 10, 50, 100],
    "kernel": ["rbf", "linear"],
    "epsilon": [0.01, 0.1, 0.2, 0.3, 0.5]
}
```

```python
model = SVR()

grid_search = GridSearchCV(model, param_grid, scoring="r2", cv=5)

grid_search.fit(X_train, y_train_scaled)
```

```python
print("best params - ", grid_search.best_params_)
```

```python
best_model = SVR(C=10, epsilon=0.1, kernel="linear")

best_model.fit(X_train, y_train_scaled)

y_test_pred_scaled = best_model.predict(X_test)
y_train_pred_scaled = best_model.predict(X_train)

print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```

# Linear SVR

```python
from sklearn.svm import LinearSVR

model = LinearSVR(C=10, epsilon=0.1, max_iter=5000)
model.fit(X_train, y_train_scaled)

y_test_pred_scaled = model.predict(X_test)
y_train_pred_scaled = model.predict(X_train)

print("Test Data R2-score: ",r2_score(y_test_scaled, y_test_pred_scaled))
print("Train Data R2-score: ",r2_score(y_train_scaled, y_train_pred_scaled))
```
