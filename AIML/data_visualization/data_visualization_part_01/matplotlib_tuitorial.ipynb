```python
# Movies Data

# Oscar Winner Movies
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

#plot
plt.plot(oscar_years,oscar_revenue)
plt.title("Oscar Years vs Revenue in($M)")
plt.xlabel("Years")
plt.ylabel("Revenue in($M)")
```

```python
# Plot multiple
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(oscar_years, oscar_revenue, label="Oscar Winners")
plt.plot(oscar_years, non_oscar_revenue, label="Non-Oscar Winners")
plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Years")
plt.ylabel("Revenue(in $M)")
plt.legend()
# plt.legend(loc="upper right") 
# plt.legend(loc="upper left")
# plt.legend(loc="lower right")
# plt.legend(loc="lower left")
# plt.legend(loc="center")
```

```python
# using Format Strings
plt.plot(oscar_years, oscar_revenue, "o-b", label="Oscar Winners")
plt.plot(oscar_years, non_oscar_revenue, ">--g", label="Non-Oscar Winners")

plt.plot(oscar_years, oscar_revenue, color="black", linestyle="-", marker="o",label="Oscar Winners")
plt.plot(oscar_years, non_oscar_revenue, color="red", linestyle="--", marker=">", linewidth=5, label="Non-Oscar Winners")
plt.legend()
```

```python
#bar chart
movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]  # in $M

plt.bar(years,revenue)

plt.title("Revenue in each year for Oscar Movies")
plt.xlabel("Years")
plt.ylabel("Revenue ($M)")

for i in range(len(years)):
    plt.text(years[i],revenue[i]+7,years[i],ha="center")  # "ha" it is for "alignment"

plt.ylim(0,max(revenue)+200)
# plt.xlim(2007,max(years)+1)   #no need by the way
```

```python
# comparing data with bar charts
oscar_movies = ["The Dark Knight","The Hurt Locker","The King's Speech","The Artist","Argo"]
oscar_revenue = [1005, 170, 427, 133, 232]  # in $M

non_oscar_movies = ["Slumdog Millionaire","Avatar","Inception","Hugo","Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

years = [2008, 2009, 2010, 2011, 2012]

# Plotting Bar Chart
import numpy as np
x = np.arange(len(years))
width = 0.4

plt.bar(x-width/2, oscar_revenue, width, label="Oscar")         # "x-width/2" Shifts to LEFT
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar") # "x-width/2" Shifts to RIGHT

plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
plt.title("Yearly revenue of Oscar vs Non-Oscar movies")
plt.legend()
plt.tight_layout()

plt.xticks(x, years) # to show x-axis labels
```
