```python
import matplotlib.pyplot as plt
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]


fig,axes=plt.subplots(2,2)

# cities_no=0
# for i in range(2):
#     for j in range(2):
#         axes[i][j].plot(days,temperatures[cities_no],marker="o")
#         axes[i][j].set_title(cities[cities_no])
#         axes[i][j].grid(True)

#         cities_no=cities_no+1


# The above big code can be made more optimal by the below code.
axes = axes.flatten() # flatten() makes the 2-D array into 1-D array

for i, ax in enumerate(axes): # enumerate() returns the index (i) and the corresponding value (ax)
    ax.plot(days, temperatures[i], marker='o')
    ax.set_title(cities[i])
    ax.grid(True)

fig.suptitle("Cities and their temp")
fig.supxlabel("Days")
fig.supylabel("Temperatures")
fig.tight_layout()
```
