```python
import matplotlib.pyplot as plt
```

```python
# Histograms

# generate random data
import numpy as np

np.random.seed(0)
# generates 100 nums with mean 70 & std dev 10
scores = np.random.normal(70, 10, 100) 
# scores = np.random.normal(loc=70, scale=10, size=100) #same


plt.hist(scores, bins=5, color="skyblue", edgecolor="black")
plt.title("Distribution of exam scores")
plt.xlabel("Marks")
plt.ylabel("Number of students")
```

```python
# custom bins
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

# custom bins
bins = [0, 30, 70, 90, 100]
plt.hist(
    marks,
    bins=bins,
    edgecolor="black",
    color="lightblue"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
```

```python
# multiple datasets
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1200.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1500, 2000, 2500, 3000, 3000
]

# Transaction amount distribution - Legit vs Fraud

plt.hist(fraud_transactions,bins=10,edgecolor="black",color="red",label="Fraud",alpha=0.5)
plt.hist(legit_transactions,bins=10,edgecolor="black",color="green",label="Legit",alpha=0.5)

plt.title("Transaction amount distribution - Legit vs Fraud")
plt.xlabel("Amount($)")
plt.ylabel("Frequency")

plt.legend()
```

```python
# axv line
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]

plt.hist(
    marks,
    bins=5,
    edgecolor="black",
    color="lightblue"
)

# line at passing marks - 33
plt.axvline(
    x=33,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Pass Mark (33)"
)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks of 30 Students")
plt.grid(True)

plt.legend()
```
