```python
import requests
```

```python
URL="https://www.scrapethissite.com/pages/simple/"
res=requests.get(URL)
```

```python
if res.status_code==200:
    print(res.headers)

with open("scraped_data/data1.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
# BeautifulSoup
from bs4 import BeautifulSoup
with open("scraped_data/data1.html","r") as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h1")
```

```python
all_h3 = soup.find_all("h3")

all_countries=[]

for h3 in all_h3:
    country_name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])   #to access attribute "class" of the parent element "div" we used "["class"]"

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    # print(h3.find_next("div").select_one("span.country-population").get_text(strip=True))

    all_countries.append([country_name, population])
    
#to access the attributes of element
print(h3.attrs)
```

```python
#storing clean data
import pandas as pd
df=pd.DataFrame(all_countries,columns=[["country_name","population"]])
df.to_csv("cleaned_data/clean_data.csv")
```
