```python
import requests
from bs4 import BeautifulSoup
```

```python
page=1
quote_author=[]
while True:
    print(f"scarping for page: {page}")
    URL=f"https://quotes.toscrape.com/page/{page}/"
    res=requests.get(URL)
    
    soup=BeautifulSoup(res.text,"lxml")
    quotes=soup.select("div.quote")

    if not quotes:
        print(f"No valid quote for page: {page}..")
        break;
        
    with open(f"scraped_data/quote{page}.html","w",encoding="utf-8") as f:
        f.write(res.text)
        print(f"Stored page:{page} successfully")
    

    quote_span=soup.find_all("span",class_="text")
    quote_span
    

    for span in quote_span:
        # each quotes
        quote=span.get_text(strip=True)
        # print(quote)

        #each author
        author=span.find_next("small",class_="author").get_text(strip=True)
        # print(author)

    
        tags_class=span.find_next("div",class_="tags")
        # print(tags_class)
    
        tag_value=tags_class.find_all("a",class_="tag")
        # print(tag_value)
    
        for tag in tag_value:
        # all tag value of individual quote
            value=tag.get_text()
            # print(value)
        
            if value=="life":
                quote_author.append([author,quote])
            else:
                continue
    page=page+1
```

```python
import pandas as pd
df=pd.DataFrame(quote_author,columns=["Author","Quote"])
df.to_csv("cleaned_data/cleandata.csv", index=False)
```
