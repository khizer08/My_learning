```python
import pandas as pd
```

```python
df=pd.read_csv("IRIS.csv")
df2=df.copy()
```

```python
# first 10 rows
df2.head(10)
# shape of dataset
df2.shape
# datatypes
df2.dtypes
```

```python
# only those rows where petal length > 4.5
df2=df2[df2["petal_length"]>4.5]
# only those rows where species is "Iris-virginica"
df2=df2[df2["species"]=="Iris-virginica"]
# both conditions
df2=df2[(df2["petal_length"]>4.5) & (df2["species"]=="Iris-virginica")]
```

```python
# Group by species and compute the "average sepal_length", "maximum petal_width", and "standard deviation of sepal_width"
df2=df.copy()
df2.groupby("species").agg(
    avg_sepal_len=("sepal_length","mean"),
    max_petal_width=("petal_width","max"),
    std_sepal_width=("sepal_width","std")
)
```

```python
# Create a new column "petal_ratio" = petal_length / petal_width. For each species, find the average petal_ratio.
df2["petal_ratio"]=df2["petal_length"]/df2["petal_width"]
df2.groupby("species")["petal_ratio"].mean()
```
