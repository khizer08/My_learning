# Prompt => LLM model => response

```python
from google import genai
from google.genai import types
import os
from dotenv import load_dotenv

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API"))
```

### Text input Text output

```python
response = client.models.generate_content(
    model="gemini-2.5-flash-lite",
    contents="a+2=7, find a",
    config=types.GenerateContentConfig(
        system_instruction="act like a math teacher"
    )
)

print(response.text)
```

### Img input Text output

```python
response = client.models.generate_content(
    model="gemini-2.5-flash-lite",
    contents=[
        types.Part.from_text(text="Answer the question, just tell the option"),
        types.Part.from_bytes(
            data=open("images/image.png", "rb").read(),
            mime_type="image/png"
        )
    ]
)

print(response.text)
```

### Personal AI Assistant

```python
class PersonalAssistant:

    def __init__(self):
        print("Hi, I am your Gemini personal assistant. How can I help?")

    def ask_question(self):

        question = input("Ask me anything: ")

        response = client.models.generate_content(
            model="gemini-2.5-flash-lite",
            contents=question,
            config=types.GenerateContentConfig(
                system_instruction="Act like a helpful personal assistant.",
                temperature=0.7,
                max_output_tokens=512
            )
        )

        print(response.text.strip())


assistant = PersonalAssistant()
assistant.ask_question()
```

### Img input Img output

```python
from huggingface_hub import InferenceClient

client = InferenceClient(
    api_key=os.getenv("HUGGING_FACE_IMG_GENERATION_API"),
    provider="auto"
)

image = client.text_to_image(
    "Generate an image of a highly detailed futuristic spaceship, flying near a nebula",
    model="black-forest-labs/FLUX.1-schnell"
)

image.save("images/generated_image.png")
```
